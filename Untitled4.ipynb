Question6: Define the terms "probability distribution" and "probability density function" (PDF). How does a  PDF differ from a probability mass function (PMF)?

Answer:

**Probability Distribution:**
A probability distribution describes how the probabilities are assigned to the possible outcomes of a random variable. It provides a full description of the likelihood of each possible value of the random variable, whether the variable is discrete or continuous.

**For a discrete random variable:** The distribution gives the probability of each specific outcome.
For a continuous random variable: The distribution assigns probabilities over intervals of values, typically described using a probability density function (PDF).

**Probability Density Function (PDF):**
A probability density function (PDF) is used for continuous random variables to describe the relative likelihood of the variable taking on a specific value. While the PDF does not give the probability of a particular value directly (since for continuous variables, the probability of any exact value is 0), it gives the probability density, which can be used to calculate the probability over an interval.

**Key points about the PDF:**

The total area under the curve of the PDF over the entire range of values is 1.
To find the probability that the variable falls within a specific interval, you compute the area under the curve of the PDF over that interval.

**Probability Mass Function (PMF):**
A probability mass function (PMF) is used for discrete random variables. It gives the probability that a discrete random variable is exactly equal to some value. Unlike a PDF, the PMF assigns specific probabilities to individual outcomes.

**Key points about the PMF:**

The sum of all probabilities in a PMF equals 1

It directly gives the probability for discrete outcomes (e.g., rolling a die or flipping a coin).

**Difference Between PDF and PMF:**
Type of Random Variable:

PDF: Used for continuous random variables (e.g., height, weight, time).

PMF: Used for discrete random variables (e.g., number of heads in coin flips, number of students in a class).
Nature of Probability:

PDF: Does not give probabilities for exact values but provides a density. The probability of the variable falling within a specific interval is found by integrating the PDF over that interval.

PMF: Directly gives the probability for each specific value.
Total Probability:

For both PDF and PMF, the total probability (sum for PMF, area under the curve for PDF) must equal 1

Question9: Given a sample mean of 105, a population mean of 100, a standard deviation of 15, and a sample  size of 25, calculate the z-score and p-value. Based on a significance level of 0.05, do you reject or fail to reject the null hypothesis?

Task: Write Python code to calculate the z-score and p-value for the given data.

Objective: Apply the formula for the z-score and interpret the p-value for hypothesis testing

```python
import math
from scipy import stats

# Given data
sample_mean = 105
population_mean = 100
std_dev = 15
sample_size = 25

# Calculate z-score
z_score = (sample_mean - population_mean) / (std_dev / math.sqrt(sample_size))

# Calculate p-value (two-tailed test)
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

# Output results
print(f"Z-Score: {z_score}")
print(f"P-Value: {p_value}")

# Decision based on significance level
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis.")
else:
    print("Fail to reject the null hypothesis.")
```

Question10: Simulate a binomial distribution with 10 trials and a probability of success of 0.6 using Python.  Generate 1,000 samples and plot the distribution. What is the expected mean and variance?

Task: Use Python to generate the data, plot the distribution, and calculate the mean and variance.

Objective: Understand the properties of a binomial distribution and verify them through simulation.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n = 10         # number of trials
p = 0.6        # probability of success
num_samples = 1000  # number of samples

# Generate binomial distribution samples
samples = np.random.binomial(n, p, num_samples)

# Calculate the expected mean and variance
expected_mean = n * p
expected_variance = n * p * (1 - p)

# Calculate the mean and variance from the simulated data
sample_mean = np.mean(samples)
sample_variance = np.var(samples)

# Plot the distribution of the samples
plt.hist(samples, bins=np.arange(0, n+2)-0.5, density=True, edgecolor='black')
plt.title('Binomial Distribution (n=10, p=0.6)')
plt.xlabel('Number of successes')
plt.ylabel('Frequency')
plt.show()

# Output results
print(f"Expected Mean: {expected_mean}")
print(f"Sample Mean: {sample_mean}")
print(f"Expected Variance: {expected_variance}")
print(f"Sample Variance: {sample_variance}")
```
